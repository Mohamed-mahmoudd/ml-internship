## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))
print("Declining pages:", (df["trend_direction"] == "down").sum())
print("Average CTR by position tier:")
print(df.groupby("position_tier")["ctr"].mean().round(4))
```

The starter dataset contains 30,000 pages,and 16,262 pages are marked as declining, which means there are many pages that may need review. Also, the average CTR changes a lot between position tiers, showing that page performance is different across search positions. These observations make this lane worth exploring.
